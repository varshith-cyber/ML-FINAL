```python
import os
import json
from collections import Counter

base = "/kaggle/input/datasets/angalakuduruvarshith/mldataproject"

original_base = os.path.join(
    base,
    "PrefEval-main (1)",
    "PrefEval-main",
    "benchmark_dataset",
    "mcq_options"
)

naturalized_path = os.path.join(
    base,
    "naturalized_independently_verified_997.json"
)

enhanced_path = os.path.join(
    base,
    "enhanced_mcq_997.json"
)

# --------------------------------------------------
# Load datasets
# --------------------------------------------------

original_all = []

for filename in os.listdir(original_base):
    if not filename.endswith(".json"):
        continue
    if filename == "travel_hotel copy.json":
        continue

    path = os.path.join(original_base, filename)

    with open(path, "r", encoding="utf-8") as f:
        data = json.load(f)

    for idx, item in enumerate(data):
        original_all.append({
            "file": filename,
            "index": idx,
            "preference": item.get("preference"),
            "question": item.get("question")
        })

with open(naturalized_path, "r", encoding="utf-8") as f:
    naturalized = json.load(f)

with open(enhanced_path, "r", encoding="utf-8") as f:
    enhanced = json.load(f)


# --------------------------------------------------
# Create matching keys
# --------------------------------------------------

def key(item):
    return (
        item.get("preference", ""),
        item.get("question", "")
    )

original_keys = {key(x) for x in original_all}

naturalized_keys = {
    (
        x.get("preference_original", ""),
        x.get("question", "")
    )
    for x in naturalized
}

enhanced_keys = {
    (
        x.get("preference", ""),
        x.get("question", "")
    )
    for x in enhanced
}


# --------------------------------------------------
# Basic counts
# --------------------------------------------------

print("=" * 70)
print("DATASET COUNTS")
print("=" * 70)

print("Original:", len(original_all))
print("Naturalized:", len(naturalized))
print("Enhanced:", len(enhanced))


# --------------------------------------------------
# Original ↔ Naturalized
# --------------------------------------------------

missing_from_original = naturalized_keys - original_keys
missing_naturalized = original_keys - naturalized_keys

print("\n" + "=" * 70)
print("ORIGINAL ↔ NATURALIZED")
print("=" * 70)

print("Naturalized examples not found in original:", len(missing_from_original))
print("Original examples missing from naturalized:", len(missing_naturalized))

if missing_naturalized:
    print("\nMissing original examples:")
    for x in list(missing_naturalized)[:10]:
        print(x)


# --------------------------------------------------
# Original ↔ Enhanced
# --------------------------------------------------

missing_from_original_enhanced = enhanced_keys - original_keys
missing_enhanced = original_keys - enhanced_keys

print("\n" + "=" * 70)
print("ORIGINAL ↔ ENHANCED")
print("=" * 70)

print("Enhanced examples not found in original:", len(missing_from_original_enhanced))
print("Original examples missing from enhanced:", len(missing_enhanced))


# --------------------------------------------------
# Naturalized ↔ Enhanced
# --------------------------------------------------

missing_from_naturalized = enhanced_keys - naturalized_keys
missing_from_enhanced = naturalized_keys - enhanced_keys

print("\n" + "=" * 70)
print("NATURALIZED ↔ ENHANCED")
print("=" * 70)

print("Enhanced examples not found in naturalized:", len(missing_from_naturalized))
print("Naturalized examples missing from enhanced:", len(missing_from_enhanced))


# --------------------------------------------------
# Enhanced valid-option distribution
# --------------------------------------------------

distribution = Counter(
    item.get("num_valid")
    for item in enhanced
)

print("\n" + "=" * 70)
print("ENHANCED VALID-OPTION DISTRIBUTION")
print("=" * 70)

for n in [1, 2, 3, 4]:
    print(f"{n} valid option(s): {distribution.get(n, 0)}")

print("Other values:", {
    k: v for k, v in distribution.items()
    if k not in [1, 2, 3, 4]
})


# --------------------------------------------------
# Validate valid_indices
# --------------------------------------------------

invalid_indices = []

for i, item in enumerate(enhanced):
    indices = item.get("valid_indices")

    if not isinstance(indices, list):
        invalid_indices.append((i, indices))
        continue

    if len(indices) != item.get("num_valid"):
        invalid_indices.append(
            (i, f"length={len(indices)}, num_valid={item.get('num_valid')}")
        )
        continue

    if any(x not in [0, 1, 2, 3] for x in indices):
        invalid_indices.append((i, indices))

print("\n" + "=" * 70)
print("VALID-INDEX CHECK")
print("=" * 70)

print("Invalid enhanced examples:", len(invalid_indices))

if invalid_indices:
    print("First problems:")
    for x in invalid_indices[:10]:
        print(x)


# --------------------------------------------------
# Duplicate checks
# --------------------------------------------------

print("\n" + "=" * 70)
print("DUPLICATE CHECK")
print("=" * 70)

print(
    "Duplicate original keys:",
    len(original_all) - len(original_keys)
)

print(
    "Duplicate naturalized keys:",
    len(naturalized) - len(naturalized_keys)
)

print(
    "Duplicate enhanced keys:",
    len(enhanced) - len(enhanced_keys)
)

print("\nAUDIT COMPLETE")
```

```python
import json
import os
from collections import defaultdict, Counter

# ============================================================
# PATHS
# ============================================================

base_path = "/kaggle/input/datasets/angalakuduruvarshith/mldataproject"

naturalized_path = os.path.join(
    base_path,
    "naturalized_independently_verified_997.json"
)

enhanced_path = os.path.join(
    base_path,
    "enhanced_mcq_997.json"
)

original_mcq_dir = os.path.join(
    base_path,
    "PrefEval-main (1)",
    "PrefEval-main",
    "benchmark_dataset",
    "mcq_options"
)

# ============================================================
# LOAD DATA
# ============================================================

with open(naturalized_path, "r", encoding="utf-8") as f:
    naturalized = json.load(f)

with open(enhanced_path, "r", encoding="utf-8") as f:
    enhanced = json.load(f)

# Load original MCQs
original = []

for filename in sorted(os.listdir(original_mcq_dir)):

    if not filename.endswith(".json"):
        continue

    # Ignore duplicate copy
    if filename == "travel_hotel copy.json":
        continue

    filepath = os.path.join(original_mcq_dir, filename)

    with open(filepath, "r", encoding="utf-8") as f:
        data = json.load(f)

    for idx, item in enumerate(data):
        original.append({
            **item,
            "_source_file": filename,
            "_source_index": idx
        })

print("Naturalized records:", len(naturalized))
print("Enhanced records:", len(enhanced))
print("Original records:", len(original))


# ============================================================
# BUILD ORIGINAL LOOKUPS
# ============================================================

# Exact source location lookup
original_by_source = {}

for item in original:
    key = (
        item["_source_file"],
        item["_source_index"]
    )

    original_by_source[key] = item


# Preference + question lookup
original_by_pq = defaultdict(list)

for item in original:

    key = (
        item.get("preference", "").strip(),
        item.get("question", "").strip()
    )

    original_by_pq[key].append(item)


# ============================================================
# BUILD ENHANCED LOOKUP
# ============================================================

# Enhanced does not have an ID.
# Use preference + question.

enhanced_by_pq = defaultdict(list)

for item in enhanced:

    key = (
        item.get("preference", "").strip(),
        item.get("question", "").strip()
    )

    enhanced_by_pq[key].append(item)


# ============================================================
# CHECK ENHANCED DUPLICATES
# ============================================================

enhanced_duplicate_keys = {
    key: items
    for key, items in enhanced_by_pq.items()
    if len(items) > 1
}

print("\nEnhanced duplicate preference/question keys:",
      len(enhanced_duplicate_keys))

if enhanced_duplicate_keys:
    print("WARNING: Enhanced dataset contains duplicate keys.")


# ============================================================
# CREATE CLEAN DATASET
# ============================================================

clean = []

missing_original = []
missing_enhanced = []
ambiguous_enhanced = []

for n_idx, n_item in enumerate(naturalized):

    # --------------------------------------------------------
    # 1. Find original using source file + source index
    # --------------------------------------------------------

    source_file = n_item.get("_source_file", n_item.get("source_file"))
    source_index = n_item.get("_source_index", n_item.get("source_index"))

    if source_file is None or source_index is None:
        missing_original.append({
            "naturalized_index": n_idx,
            "reason": "Missing source_file/source_index"
        })
        continue

    original_key = (
        source_file,
        source_index
    )

    original_item = original_by_source.get(original_key)

    if original_item is None:
        missing_original.append({
            "naturalized_index": n_idx,
            "source_file": source_file,
            "source_index": source_index
        })
        continue


    # --------------------------------------------------------
    # 2. Find enhanced using preference + question
    # --------------------------------------------------------

    preference = original_item.get("preference", "").strip()
    question = original_item.get("question", "").strip()

    pq_key = (preference, question)

    enhanced_matches = enhanced_by_pq.get(pq_key, [])

    if len(enhanced_matches) == 0:
        missing_enhanced.append({
            "naturalized_index": n_idx,
            "source_file": source_file,
            "source_index": source_index,
            "preference": preference,
            "question": question
        })
        continue

    if len(enhanced_matches) > 1:
        ambiguous_enhanced.append({
            "naturalized_index": n_idx,
            "source_file": source_file,
            "source_index": source_index,
            "matches": len(enhanced_matches)
        })
        continue

    enhanced_item = enhanced_matches[0]


    # --------------------------------------------------------
    # 3. Build clean record
    # --------------------------------------------------------

    clean_item = {
        "id": f"prefeval_{n_idx:04d}",

        "source_file": source_file,
        "source_index": source_index,

        "original_preference": original_item.get("preference"),
        "naturalized_preference": (
            n_item.get("preference_final_independent")
            or n_item.get("preference_final")
            or n_item.get("preference_naturalized")
        ),

        "question": question,

        # Original MCQ
        "original_options": original_item.get(
            "classification_task_options"
        ),

        # Enhanced MCQ
        "enhanced_options": enhanced_item.get("options"),

        # Ground truth for enhanced MCQ
        "valid_indices": enhanced_item.get("valid_indices"),

        "num_valid": enhanced_item.get("num_valid"),

        # Keep original information for later C1/C2 scoring
        "original_explanation": original_item.get("explanation"),
        "original_violation_probability": original_item.get(
            "violation_probability"
        )
    }

    clean.append(clean_item)


# ============================================================
# REPORT
# ============================================================

print("\n" + "=" * 60)
print("CLEAN DATASET AUDIT")
print("=" * 60)

print("Clean records:", len(clean))
print("Missing original:", len(missing_original))
print("Missing enhanced:", len(missing_enhanced))
print("Ambiguous enhanced:", len(ambiguous_enhanced))


# Enhanced valid-option distribution

valid_distribution = Counter(
    item["num_valid"]
    for item in clean
)

print("\nEnhanced valid-option distribution:")

for k in sorted(valid_distribution):
    print(
        f"{k} valid option(s): "
        f"{valid_distribution[k]}"
    )


# ============================================================
# SAVE
# ============================================================

output_path = "/kaggle/working/prefeval_clean_997.json"

with open(output_path, "w", encoding="utf-8") as f:
    json.dump(
        clean,
        f,
        indent=2,
        ensure_ascii=False
    )

print("\nSaved to:")
print(output_path)


# ============================================================
# FINAL SAFETY CHECK
# ============================================================

assert len(clean) == 997, \
    f"Expected 997 clean records, got {len(clean)}"

assert len(missing_original) == 0, \
    f"Missing original records: {len(missing_original)}"

assert len(missing_enhanced) == 0, \
    f"Missing enhanced records: {len(missing_enhanced)}"

assert len(ambiguous_enhanced) == 0, \
    f"Ambiguous enhanced records: {len(ambiguous_enhanced)}"

print("\n✅ ALL 997 RECORDS ALIGNED SUCCESSFULLY")
```

```python
import json
import glob
import os
from collections import Counter

# ============================================================
# PATHS
# ============================================================

BASE = "/kaggle/input/datasets/angalakuduruvarshith/mldataproject"

NATURALIZED_PATH = os.path.join(
    BASE,
    "naturalized_independently_verified_997.json"
)

ENHANCED_PATH = os.path.join(
    BASE,
    "enhanced_mcq_997.json"
)

MCQ_DIR = os.path.join(
    BASE,
    "PrefEval-main (1)",
    "PrefEval-main",
    "benchmark_dataset",
    "mcq_options"
)

# ============================================================
# LOAD DATA
# ============================================================

with open(NATURALIZED_PATH, "r", encoding="utf-8") as f:
    naturalized = json.load(f)

with open(ENHANCED_PATH, "r", encoding="utf-8") as f:
    enhanced = json.load(f)

# ============================================================
# LOAD ORIGINAL MCQs
# ============================================================

original = []

mcq_files = sorted(
    f for f in glob.glob(os.path.join(MCQ_DIR, "*.json"))
    if os.path.basename(f).lower() != "travel_hotel copy.json"
)

for file_path in mcq_files:
    with open(file_path, "r", encoding="utf-8") as f:
        records = json.load(f)

    source_file = os.path.basename(file_path)

    for idx, record in enumerate(records):
        original.append({
            **record,
            "_source_file": source_file,
            "_source_index": idx
        })

print("Naturalized:", len(naturalized))
print("Enhanced:", len(enhanced))
print("Original:", len(original))

# ============================================================
# INDEX ORIGINAL
# ============================================================

original_lookup = {
    (r["_source_file"], r["_source_index"]): r
    for r in original
}

# ============================================================
# INDEX ENHANCED
# ============================================================

enhanced_lookup = {
    (r["preference"], r["question"]): r
    for r in enhanced
}

# ============================================================
# BUILD FINAL DATASET
# ============================================================

clean = []

missing_original = []
missing_enhanced = []

for n in naturalized:

    source_file = n.get("_source_file")
    source_index = n.get("_source_index")

    original_key = (source_file, source_index)

    o = original_lookup.get(original_key)

    if o is None:
        missing_original.append(original_key)
        continue

    enhanced_key = (
        o["preference"],
        o["question"]
    )

    e = enhanced_lookup.get(enhanced_key)

    if e is None:
        missing_enhanced.append(enhanced_key)
        continue

    # --------------------------------------------------------
    # ORIGINAL PREFEVAL GROUND TRUTH
    #
    # Original PREFEVAL source code states:
    # first option in the JSON is the correct answer.
    # Therefore: A
    # --------------------------------------------------------

    original_correct = "A"

    # --------------------------------------------------------
    # ENHANCED GROUND TRUTH
    # --------------------------------------------------------

    enhanced_correct = sorted(
        chr(65 + i)
        for i in e["valid_indices"]
    )

    # Use source file + index as stable ID
    record_id = f"{source_file}__{source_index}"

    clean.append({
        "id": record_id,

        "source_file": source_file,
        "source_index": source_index,

        "original_preference": o["preference"],
        "naturalized_preference": n["preference_final"],

        "question": o["question"],

        "original_options": o["classification_task_options"],
        "enhanced_options": e["options"],

        "original_correct": original_correct,

        "enhanced_correct": enhanced_correct,
        "valid_indices": e["valid_indices"],
        "num_valid": len(enhanced_correct),

        "original_explanation": o.get("explanation"),
        "original_violation_probability": o.get(
            "violation_probability"
        )
    })

# ============================================================
# AUDIT
# ============================================================

print("\n" + "=" * 60)
print("FINAL EVALUATION DATASET AUDIT")
print("=" * 60)

print("Clean records:", len(clean))
print("Missing original:", len(missing_original))
print("Missing enhanced:", len(missing_enhanced))

print("\nOriginal ground truth:")
print(
    Counter(
        r["original_correct"]
        for r in clean
    )
)

print("\nEnhanced valid-option distribution:")

dist = Counter(
    r["num_valid"]
    for r in clean
)

for n in [1, 2, 3, 4]:
    print(f"{n} valid option(s): {dist[n]}")

# ============================================================
# CHECK VALID INDICES
# ============================================================

invalid = []

for r in clean:
    if any(
        i not in [0, 1, 2, 3]
        for i in r["valid_indices"]
    ):
        invalid.append(r["id"])

print("\nInvalid enhanced indices:", len(invalid))

# ============================================================
# SAVE
# ============================================================

FINAL_PATH = "/kaggle/working/prefeval_evaluation_997.json"

with open(FINAL_PATH, "w", encoding="utf-8") as f:
    json.dump(
        clean,
        f,
        indent=2,
        ensure_ascii=False
    )

print("\nSaved:")
print(FINAL_PATH)

# ============================================================
# SHOW FIRST 3
# ============================================================

print("\n" + "=" * 60)
print("FIRST 3 RECORDS")
print("=" * 60)

for r in clean[:3]:

    print("\nID:", r["id"])
    print("Original correct:", r["original_correct"])
    print("Enhanced correct:", r["enhanced_correct"])
    print("Num valid:", r["num_valid"])

    print("\nEnhanced options:")

    for i, option in enumerate(r["enhanced_options"]):
        letter = chr(65 + i)

        marker = (
            " <-- CORRECT"
            if letter in r["enhanced_correct"]
            else ""
        )

        print(f"  {letter}. {option}{marker}")
```

```python
def ask_model(preference, question, options, multi_label=False):
    option_text = "\n".join(
        f"{chr(65+i)}. {opt}"
        for i, opt in enumerate(options)
    )

    if multi_label:
        instruction = (
            "Select ALL options that are valid according to the preference. "
            "There may be 1, 2, 3, or 4 correct options. "
            "Return ONLY the option letters separated by commas. "
            "Do not provide an explanation. "
            "Example: A,C,D"
        )
    else:
        instruction = (
            "Select the ONE option that best matches the preference. "
            "Return ONLY one option letter. "
            "Do not provide an explanation. "
            "Example: B"
        )

    prompt = f"""Preference:
{preference}

Question:
{question}

Options:
{option_text}

{instruction}
"""

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "system",
                "content": (
                    "You are evaluating preference-following "
                    "in multiple-choice questions."
                )
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0,
        max_tokens=50
    )

    return response.choices[0].message.content.strip()


def parse_single(response):
    """
    Parse one answer for original single-label MCQs.
    """
    if not response:
        return None

    letters = re.findall(
        r'\b([ABCD])\b',
        response.upper()
    )

    return letters[0] if letters else None


def parse_multi(response):
    """
    Parse all A/B/C/D answers for enhanced multi-label MCQs.
    """
    if not response:
        return []

    letters = re.findall(
        r'\b([ABCD])\b',
        response.upper()
    )

    return sorted(set(letters))
```

```python
import json
import random
from collections import Counter

INPUT = "/kaggle/working/prefeval_evaluation_997.json"
OUTPUT = "/kaggle/working/prefeval_evaluation_shuffled_997.json"

with open(INPUT, "r", encoding="utf-8") as f:
    data = json.load(f)

# Fixed seed = reproducible shuffling
SEED = 20261030
rng = random.Random(SEED)

shuffled_data = []

for r in data:
    item = dict(r)

    # ============================================================
    # ORIGINAL MCQ
    # Original correct answer is always the FIRST option
    # ============================================================
    original_options = list(r["original_options"])
    original_correct_text = original_options[0]

    indices = list(range(len(original_options)))
    rng.shuffle(indices)

    shuffled_original = [
        original_options[i] for i in indices
    ]

    new_original_correct_idx = shuffled_original.index(
        original_correct_text
    )

    item["original_options_shuffled"] = shuffled_original
    item["original_correct_shuffled"] = chr(
        65 + new_original_correct_idx
    )

    # ============================================================
    # ENHANCED MCQ
    # valid_indices refer to the ORIGINAL enhanced option order
    # ============================================================
    enhanced_options = list(r["enhanced_options"])
    valid_indices = set(r["valid_indices"])

    indices = list(range(len(enhanced_options)))
    rng.shuffle(indices)

    shuffled_enhanced = [
        enhanced_options[i] for i in indices
    ]

    # Convert old valid indices -> new shuffled positions
    new_valid_positions = sorted(
        indices.index(i)
        for i in valid_indices
    )

    new_enhanced_correct = [
        chr(65 + i)
        for i in new_valid_positions
    ]

    item["enhanced_options_shuffled"] = shuffled_enhanced
    item["enhanced_correct_shuffled"] = new_enhanced_correct
    item["enhanced_num_valid"] = len(new_enhanced_correct)

    shuffled_data.append(item)


with open(OUTPUT, "w", encoding="utf-8") as f:
    json.dump(shuffled_data, f, indent=2, ensure_ascii=False)

print("=" * 70)
print("SHUFFLED DATASET CREATED")
print("=" * 70)
print("Records:", len(shuffled_data))
print("Seed:", SEED)
print("Saved:", OUTPUT)

print("\nOriginal shuffled answer distribution:")
print(
    Counter(
        r["original_correct_shuffled"]
        for r in shuffled_data
    )
)

print("\nEnhanced valid-option distribution:")
print(
    Counter(
        r["enhanced_num_valid"]
        for r in shuffled_data
    )
)
```

```python
import os
import json
import re
import time
from datetime import datetime
from groq import Groq
from kaggle_secrets import UserSecretsClient

# ============================================================
# CONFIG
# ============================================================

INPUT_FILE = "/kaggle/working/prefeval_evaluation_shuffled_997.json"

CHECKPOINT_FILE = "/kaggle/working/prefeval_full_results_checkpoint.json"
FINAL_RESULTS_FILE = "/kaggle/working/prefeval_full_results_final.json"
METRICS_FILE = "/kaggle/working/prefeval_metrics.json"

MODEL = "openai/gpt-oss-20b"

MAX_RETRIES = 5
RETRY_BASE_SECONDS = 5

# ============================================================
# LOAD API KEY
# ============================================================

secrets = UserSecretsClient()

api_key = secrets.get_secret("apikey")
os.environ["GROQ_API_KEY"] = api_key

client = Groq(api_key=os.environ["GROQ_API_KEY"])

print("API key loaded:", bool(api_key))
print("Model:", MODEL)

# ============================================================
# LOAD DATA
# ============================================================

with open(INPUT_FILE, "r", encoding="utf-8") as f:
    data = json.load(f)

print("Total examples:", len(data))

# ============================================================
# MODEL CALL
# ============================================================

def ask_model(preference, question, options, multi_label=False):

    option_text = "\n".join(
        f"{chr(65+i)}. {opt}"
        for i, opt in enumerate(options)
    )

    if multi_label:
        instruction = (
            "Select ALL options that are valid according to the preference. "
            "There may be 1, 2, 3, or 4 correct options. "
            "Return ONLY the option letters separated by commas. "
            "Do not provide an explanation. "
            "Example: A,C,D"
        )
    else:
        instruction = (
            "Select the ONE option that best matches the preference. "
            "Return ONLY one option letter. "
            "Do not provide an explanation. "
            "Example: B"
        )

    prompt = f"""Preference:
{preference}

Question:
{question}

Options:
{option_text}

{instruction}
"""

    for attempt in range(1, MAX_RETRIES + 1):

        try:

            response = client.chat.completions.create(
                model=MODEL,
                messages=[
                    {
                        "role": "system",
                        "content": (
                            "You are evaluating preference-following in "
                            "multiple-choice questions. "
                            "Think internally, but your final answer must "
                            "contain only the requested option letters."
                        )
                    },
                    {
                        "role": "user",
                        "content": prompt
                    }
                ],
                temperature=0,
                max_tokens=300
            )

            message = response.choices[0].message

            content = message.content

            if content and content.strip():
                return content.strip()

            # Fallback if model returns empty final content
            reasoning = getattr(message, "reasoning", None)

            if reasoning and reasoning.strip():
                return reasoning.strip()

            return ""

        except Exception as e:

            print(
                f"      API error "
                f"(attempt {attempt}/{MAX_RETRIES}): {e}"
            )

            if attempt < MAX_RETRIES:

                wait_time = RETRY_BASE_SECONDS * (2 ** (attempt - 1))

                print(
                    f"      Waiting {wait_time}s before retry..."
                )

                time.sleep(wait_time)

            else:

                raise e


# ============================================================
# PARSERS
# ============================================================

def parse_single(response):

    if not response:
        return None

    # Prefer a clean standalone answer
    text = response.strip().upper()

    match = re.fullmatch(r"\s*([ABCD])\s*[.)]?\s*", text)

    if match:
        return match.group(1)

    # Fallback
    letters = re.findall(r"\b([ABCD])\b", text)

    return letters[0] if letters else None


def parse_multi(response):

    if not response:
        return []

    text = response.strip().upper()

    # Handle normal formats:
    # A,C,D
    # A, C, D
    # A C D
    # A,B
    letters = re.findall(r"\b([ABCD])\b", text)

    return sorted(set(letters))


# ============================================================
# CHECKPOINT FUNCTIONS
# ============================================================

def load_checkpoint():

    if not os.path.exists(CHECKPOINT_FILE):
        return {}

    try:

        with open(CHECKPOINT_FILE, "r", encoding="utf-8") as f:
            checkpoint = json.load(f)

        print(
            f"\nExisting checkpoint found: "
            f"{len(checkpoint)} completed examples"
        )

        return checkpoint

    except Exception as e:

        print("Could not load checkpoint:", e)
        return {}


def save_checkpoint(results):

    temp_file = CHECKPOINT_FILE + ".tmp"

    with open(temp_file, "w", encoding="utf-8") as f:
        json.dump(
            results,
            f,
            indent=2,
            ensure_ascii=False
        )

    # Atomic replacement
    os.replace(temp_file, CHECKPOINT_FILE)


# ============================================================
# LOAD EXISTING CHECKPOINT
# ============================================================

results = load_checkpoint()

completed_ids = set(results.keys())

print(
    f"Already completed: "
    f"{len(completed_ids)}/{len(data)}"
)

remaining = [
    r for r in data
    if r["id"] not in completed_ids
]

print(
    f"Remaining examples: {len(remaining)}"
)

print("=" * 70)


# ============================================================
# FULL EVALUATION
# ============================================================

start_time = time.time()

for position, r in enumerate(remaining, 1):

    example_id = r["id"]

    print(
        f"\nRunning "
        f"{position}/{len(remaining)} "
        f"(overall {len(completed_ids)+1}/{len(data)}): "
        f"{example_id}"
    )

    # --------------------------------------------------------
    # C1
    # --------------------------------------------------------

    print("   C1...", end=" ", flush=True)

    c1_raw = ask_model(
        r["original_preference"],
        r["question"],
        r["original_options_shuffled"],
        multi_label=False
    )

    c1_pred = parse_single(c1_raw)

    print(c1_pred)

    # --------------------------------------------------------
    # C2
    # --------------------------------------------------------

    print("   C2...", end=" ", flush=True)

    c2_raw = ask_model(
        r["naturalized_preference"],
        r["question"],
        r["original_options_shuffled"],
        multi_label=False
    )

    c2_pred = parse_single(c2_raw)

    print(c2_pred)

    # --------------------------------------------------------
    # C3
    # --------------------------------------------------------

    print("   C3...", end=" ", flush=True)

    c3_raw = ask_model(
        r["original_preference"],
        r["question"],
        r["enhanced_options_shuffled"],
        multi_label=True
    )

    c3_pred = parse_multi(c3_raw)

    print(c3_pred)

    # --------------------------------------------------------
    # C4
    # --------------------------------------------------------

    print("   C4...", end=" ", flush=True)

    c4_raw = ask_model(
        r["naturalized_preference"],
        r["question"],
        r["enhanced_options_shuffled"],
        multi_label=True
    )

    c4_pred = parse_multi(c4_raw)

    print(c4_pred)

    # --------------------------------------------------------
    # GROUND TRUTH
    # --------------------------------------------------------

    c1_gt = r["original_correct_shuffled"]

    c2_gt = r["original_correct_shuffled"]

    c3_gt = r["enhanced_correct_shuffled"]

    c4_gt = r["enhanced_correct_shuffled"]

    # --------------------------------------------------------
    # SAVE COMPLETE EXAMPLE
    # --------------------------------------------------------

    results[example_id] = {
        "id": example_id,

        "c1": {
            "preference_type": "original",
            "mcq_type": "original",
            "raw_response": c1_raw,
            "prediction": c1_pred,
            "ground_truth": c1_gt,
            "correct": c1_pred == c1_gt
        },

        "c2": {
            "preference_type": "naturalized",
            "mcq_type": "original",
            "raw_response": c2_raw,
            "prediction": c2_pred,
            "ground_truth": c2_gt,
            "correct": c2_pred == c2_gt
        },

        "c3": {
            "preference_type": "original",
            "mcq_type": "enhanced",
            "raw_response": c3_raw,
            "prediction": c3_pred,
            "ground_truth": c3_gt,
            "exact_match": set(c3_pred) == set(c3_gt)
        },

        "c4": {
            "preference_type": "naturalized",
            "mcq_type": "enhanced",
            "raw_response": c4_raw,
            "prediction": c4_pred,
            "ground_truth": c4_gt,
            "exact_match": set(c4_pred) == set(c4_gt)
        },

        "num_valid": r["enhanced_num_valid"],

        "completed_at": datetime.now().isoformat()
    }

    # --------------------------------------------------------
    # SAVE IMMEDIATELY
    # --------------------------------------------------------

    save_checkpoint(results)

    completed_ids.add(example_id)

    print(
        f"   ✓ SAVED "
        f"({len(completed_ids)}/{len(data)})"
    )


# ============================================================
# FINAL RESULTS
# ============================================================

print("\n" + "=" * 70)
print("ALL EXAMPLES COMPLETED")
print("=" * 70)

with open(FINAL_RESULTS_FILE, "w", encoding="utf-8") as f:
    json.dump(
        results,
        f,
        indent=2,
        ensure_ascii=False
    )

print("Saved:")
print(FINAL_RESULTS_FILE)
```

```python
import json
import numpy as np
from collections import defaultdict

FILE = "/kaggle/working/prefeval_full_results_final.json"

with open(FILE, "r", encoding="utf-8") as f:
    results = json.load(f)

N = len(results)

# Counters
c1_correct = 0
c2_correct = 0
c3_exact = 0
c4_exact = 0

c3_precision = []
c3_recall = []
c3_f1 = []

c4_precision = []
c4_recall = []
c4_f1 = []

# Breakdown by number of valid options
breakdown = defaultdict(lambda: {
    "total": 0,
    "c3_exact": 0,
    "c4_exact": 0,
    "c3_precision": [],
    "c3_recall": [],
    "c3_f1": [],
    "c4_precision": [],
    "c4_recall": [],
    "c4_f1": []
})


def set_metrics(pred, truth):
    pred = set(pred)
    truth = set(truth)

    intersection = len(pred & truth)

    precision = intersection / len(pred) if pred else 0
    recall = intersection / len(truth) if truth else 0

    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall > 0 else 0
    )

    exact = pred == truth

    return exact, precision, recall, f1


# =========================================================
# PROCESS ALL 997 RESULTS
# =========================================================

for r in results.values():

    # -------------------------
    # C1
    # -------------------------
    if r["c1"]["correct"]:
        c1_correct += 1

    # -------------------------
    # C2
    # -------------------------
    if r["c2"]["correct"]:
        c2_correct += 1

    # -------------------------
    # C3
    # -------------------------
    pred3 = r["c3"]["prediction"]
    truth3 = r["c3"]["ground_truth"]

    exact3, p3, rec3, f3 = set_metrics(pred3, truth3)

    if exact3:
        c3_exact += 1

    c3_precision.append(p3)
    c3_recall.append(rec3)
    c3_f1.append(f3)

    # -------------------------
    # C4
    # -------------------------
    pred4 = r["c4"]["prediction"]
    truth4 = r["c4"]["ground_truth"]

    exact4, p4, rec4, f4 = set_metrics(pred4, truth4)

    if exact4:
        c4_exact += 1

    c4_precision.append(p4)
    c4_recall.append(rec4)
    c4_f1.append(f4)

    # -------------------------
    # Breakdown
    # -------------------------
    k = r["num_valid"]

    b = breakdown[k]
    b["total"] += 1

    if exact3:
        b["c3_exact"] += 1

    if exact4:
        b["c4_exact"] += 1

    b["c3_precision"].append(p3)
    b["c3_recall"].append(rec3)
    b["c3_f1"].append(f3)

    b["c4_precision"].append(p4)
    b["c4_recall"].append(rec4)
    b["c4_f1"].append(f4)


# =========================================================
# FINAL RESULTS
# =========================================================

print("=" * 70)
print("FINAL PREFEVAL EVALUATION RESULTS")
print("=" * 70)

print(f"\nTotal Examples: {N}")

print("\nC1 — Original Preference → Original MCQ")
print(f"Exact Accuracy : {100*c1_correct/N:.2f}%")
print(f"Correct        : {c1_correct}/{N}")

print("\nC2 — Naturalized Preference → Original MCQ")
print(f"Exact Accuracy : {100*c2_correct/N:.2f}%")
print(f"Correct        : {c2_correct}/{N}")

print("\nC3 — Original Preference → Enhanced MCQ")
print(f"Exact Set Match: {100*c3_exact/N:.2f}%")
print(f"Precision      : {100*np.mean(c3_precision):.2f}%")
print(f"Recall         : {100*np.mean(c3_recall):.2f}%")
print(f"F1 Score       : {100*np.mean(c3_f1):.2f}%")

print("\nC4 — Naturalized Preference → Enhanced MCQ")
print(f"Exact Set Match: {100*c4_exact/N:.2f}%")
print(f"Precision      : {100*np.mean(c4_precision):.2f}%")
print(f"Recall         : {100*np.mean(c4_recall):.2f}%")
print(f"F1 Score       : {100*np.mean(c4_f1):.2f}%")


# =========================================================
# BREAKDOWN BY VALID OPTION COUNT
# =========================================================

print("\n" + "=" * 70)
print("PERFORMANCE BY NUMBER OF VALID OPTIONS")
print("=" * 70)

for k in sorted(breakdown):

    b = breakdown[k]
    total = b["total"]

    print(f"\n{k} VALID OPTION(S) — {total} examples")

    print(
        f"C3 | Exact: {100*b['c3_exact']/total:.2f}% | "
        f"P: {100*np.mean(b['c3_precision']):.2f}% | "
        f"R: {100*np.mean(b['c3_recall']):.2f}% | "
        f"F1: {100*np.mean(b['c3_f1']):.2f}%"
    )

    print(
        f"C4 | Exact: {100*b['c4_exact']/total:.2f}% | "
        f"P: {100*np.mean(b['c4_precision']):.2f}% | "
        f"R: {100*np.mean(b['c4_recall']):.2f}% | "
        f"F1: {100*np.mean(b['c4_f1']):.2f}%"
    )
```
